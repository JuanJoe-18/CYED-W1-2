## Seccion A: Cheat Sheet

```python
import re

# OPERACIONES CLAVE:
# re.match(pattern, string)     -> match al INICIO de la cadena
# re.search(pattern, string)    -> match en CUALQUIER posicion
# re.fullmatch(pattern, string) -> match de la cadena COMPLETA
# re.findall(pattern, string)   -> lista de TODOS los matches
# re.finditer(pattern, string)  -> iterador de match objects
# re.sub(pattern, repl, string) -> reemplazar matches
# re.split(pattern, string)     -> dividir en matches

print("re module loaded:", re.__name__)
```

### Clases de caracteres y cuantificadores

| Patron | Significado |
|--------|-------------|
| `\d` / `\D` | digito / no-digito |
| `\w` / `\W` | word char / no-word char |
| `\s` / `\S` | whitespace / no-whitespace |
| `[abc]` | uno de a, b, o c |
| `[^abc]` | cualquiera excepto a, b, c |
| `?` | 0 o 1 |
| `*` | 0 o mas |
| `+` | 1 o mas |
| `{n,m}` | entre n y m |
| `()` | grupo |

alternativas: `|` (ejemplo: `a|b` = a o b)

```python
text = "Version 3.1 costs $500"
print("digits:", re.findall(r'\d+', text))
print("words:", re.findall(r'\w+', text))
print("literal $:", re.findall(r'\$', text))
```

### Grupos de captura
- `(pattern)` - grupo capturante (numerado 1, 2, 3...)
- `(?:pattern)` - grupo NO capturante (solo agrupa)
- `(?P<name>pattern)` - grupo nombrado
- `findall` con grupos -> devuelve tuplas de grupos

```python
date_p = re.compile(r'(\d{4})-(\d{2})-(\d{2})')
m = date_p.search("Date: 2026-08-24")
if m:
    print("groups:", m.groups())
    print("group 1:", m.group(1))

time_p = re.compile(r'(?P<hour>\d{2}):(?P<min>\d{2})')
m = time_p.search("at 14:30")
if m:
    print("named:", m.groupdict())
```

### Lookaheads, Lookbehinds y Backreferences
- `(?=...)` - positive lookahead (lo que sigue DEBE ser...)
- `(?!...)` - negative lookahead (lo que sigue NO DEBE ser...)
- `(?<=...)` - positive lookbehind (lo anterior DEBE ser...)
- `(?<!...)` - negative lookbehind (lo anterior NO DEBE ser...)
- `\1` - backreference al grupo 1

```python
# Lookahead: word followed by a digit
print("lookahead:", re.findall(r'\w+(?=\d)', ["abc", "abc1", "xyz9"]))

# Lookbehind: @ not preceded by word char
print("lookbehind:", re.findall(r'(?<!\w)@user', ["@user", "email@user"]))

# Backreference: same separator both times
phone = re.compile(r'\d{3}([ -]?)\d{3}\1\d{3}')
print("same sep:", phone.fullmatch("555-555-555"))
print("diff sep:", phone.fullmatch("555-555 555"))
```

### Boundaries
- `^` - inicio de linea
- `$` - fin de linea
- `\b` - word boundary
- `\B` - non-word boundary
- `\A` - inicio de string
- `\Z` - fin de string

Ejercicios seleccionados que entrenan las habilidades necesarias para los 5 ejercicios del seguimiento.

```python
# WARM-UP 1: findall, sub, split
text = "cats :3, dogs :5, birds :2"

print("1. Get all numbers:", re.findall(r'\d+', text))
print("2. Replace : with =:", re.sub(r':', ' = ', text))
print("3. Split on ',':", re.split(r', ', text))
```

```python
# WARM-UP 2: Word boundaries - replace only whole word 'red'
text = 'bred red spread credible red.'
result = re.sub(r'\bred\b', 'brown', text)
print(result)
assert result == 'bred brown spread credible brown.'
```

```python
# WARM-UP 3: Filter words containing BOTH 'e' AND 'n'
words = ['fast', 'some', 'wonderful', 'army', 'wally', 'waterpolo', 'food', 'eat', 'new', 'corn']
pattern = r'e.*n|n.*e'
result = [w for w in words if re.search(pattern, w)]
print(result)
assert result == ['wonderful', 'new']
```

```python
# WARM-UP 4: Replace only the FIRST occurrence of '5' with 'five'
lines = ['They run 5 kilometers in 5 minutes',
         'They were only 5 girls',
         'We are eating 5 apples and 5 oranges']
result = [re.sub(r'5', 'five', line, count=1) for line in lines]
print(result)
```

```python
# WARM-UP 5: Filter elements containing 42 surrounded by word characters
words = ['hi42bye', 'nice1423', 'bad42', 'cool_42a', '42fake', '_42_', 'lucky42man']
pattern = r'\w42\w'
result = [w for w in words if re.search(pattern, w)]
print(result)
assert result == ['hi42bye', 'nice1423', 'cool_42a', '_42_', 'lucky42man']
```

### 1) [20 pts] Standardizing ISO Timestamps
**Requirements:**
- Format: YYYY-MM-DDTHH:MM:SS
- Month: 01-12
- Day: 01-31
- Capture Year, Month, Day as separate groups

```python
def parse_iso_date(timestamp):
    pattern = r'^(\d{4})-(0[1-9]|1[0-2])-(0[1-9]|[12]\d|3[01])T'  
    match = re.match(pattern, timestamp)
    return match.groups() if match else None

test_cases_5 = [
    ("2023-05-15T14:30:05", ('2023', '05', '15')),
    ("1999-12-31T23:59:59", ('1999', '12', '31')),
    ("2023-13-01T14:30:00", None),
    ("2023-01-32T14:30:00", None),
    ("2023-00-15T14:30:00", None),
    ("23-05-15T14:30:00", None),
    ("2023-5-15T14:30:00", None),
    ("2023-05-15 14:30:00", None),
    ("2023-02-28T10:00:00", ('2023', '02', '28')),
    ("2023-06-31T10:00:00", ('2023', '06', '31')),
]

for data, expected in test_cases_5:
    result = parse_iso_date(data)
    print(f"{data!r:30} -> {result}")
    assert result == expected, f"FAIL: {data!r} expected {expected}, got {result}"
print("\nAll test cases passed!")
```

### 2) [20 pts] Validating Product SKU Patterns
**Requirements:**
- 2 uppercase letters (Department Code)
- hyphen
- 4 digits (Year Code)
- hyphen
- Serial: at least 3 alphanumeric, last char must be a digit

```python
def is_valid_sku(sku):
    pattern = r'^([A-Z]{2})-(\d{4})-([A-Za-z\d]{2,}\d)$'
    return bool(re.match(pattern, sku))

test_cases_2 = [
    ("EL-2023-A109", True),
    ("TS-1998-999ZZ0", True),
    ("el-2023-A109", False),
    ("E1-2023-A109", False),
    ("EL-23-A109", False),
    ("EL-2023-ABC", False),
    ("EL-2023-A", False),
    ("EL2023-A109", False),
    ("EL-2023-A109 ", False),
    ("-EL-2023-A109", False),
]

for data, expected in test_cases_2:
    result = is_valid_sku(data)
    print(f"{data!r:25} -> {result}")
    assert result == expected, f"FAIL: {data!r} expected {expected}, got {result}"
print("\nAll test cases passed!")
```

### 4) [20 pts] Cleaning Social Media Mentions & Hashtags
**Requirements:**
- Mentions: @ + letters, digits, underscores
- Hashtags: # + must contain at least one letter (not just numbers)
- Use lookaheads for hashtag validation

```python
def get_social_tags(tweet):
    pattern = r'(?!)'  # TODO: Tu patron aqui
    return re.findall(pattern, tweet)

test_cases_3 = [
    ("#Python", ["#Python"]),
    ("@Guido", ["@Guido"]),
    ("#12345", []),
    ("#Data_Science", ["#Data_Science"]),
    ("@user123", ["@user123"]),
    ("Check #1", []),
    ("#A1", ["#A1"]),
    ("email@domain.com", []),
    ("#_underscore", ["#_underscore"]),
    ("#Regex2026", ["#Regex2026"]),
]

for data, expected in test_cases_3:
    result = get_social_tags(data)
    print(f"{data!r:30} -> {result}")
    assert result == expected, f"FAIL: {data!r} expected {expected}, got {result}"
print("\nAll test cases passed!")
```

### 5) [20 pts] Parsing SQL Table Aliases
**Requirements:**
- Find FROM [table] AS [alias]
- Tables and aliases: alphanumeric + underscores
- Table name != alias (use backreferences + negative lookahead)

```python
def find_table_aliases(query):
    pattern = r'(?!)'  # TODO: Tu patron aqui
    return re.findall(pattern, query, re.IGNORECASE)

test_cases_4 = [
    ("FROM users AS u", [('users', 'u')]),
    ("FROM Orders AS o", [('Orders', 'o')]),
    ("FROM logs AS logs", []),
    ("FROM products AS p JOIN sales AS s", [('products', 'p'), ('sales', 's')]),
    ("FROM staff AS staff_member", [('staff', 'staff_member')]),
    ("FROM 123table AS t", [('123table', 't')]),
    ("FROM users AS ", []),
    ("SELECT * FROM users", []),
    ("from assets as a", [('assets', 'a')]),
    ("FROM schema.table AS t", []),
    ("join loans AS l", []),
]

for data, expected in test_cases_4:
    result = find_table_aliases(data)
    print(f"{data!r:45} -> {result}")
    assert result == expected, f"FAIL: {data!r} expected {expected}, got {result}"
print("\nAll test cases passed!")
```
