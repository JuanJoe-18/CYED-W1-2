```python
text = "cats :3, dogs :5, birds :2"

# 1. Get all the numbers with re.findall()
# 2. Replace every : with " = " using re.sub()
# 3. Break the string into parts with re.split(), splitting on ", "
```

```python
# first
print(re.findall(r"\d+", text))
# second
print(re.sub(r":", " = ", text))
# third one
print(re.split(r", ", text))
```

```python
text2 = "Room_12B has 3 windows!"

# 1. Find every word-character token: \w+
# 2. Find every digit run: \d+
# 3. Find every run of characters that are not word characters: \W+

# \w+
re.findall(r"\w+", text2)
# \d+
re.findall(r"\d+", text2)
# \W+
re.findall(r"\W+", text2)
```

```python
re.findall(r"[^aeiou]", "hello world")
```

```python
patroncito = r"[A-Z]\d+"

stringsito = "Mi codigo es Ñ00412281"
re.findall(patroncito, stringsito)
```

```python
receipt = "Item1 costs $20 , Ttem_2 Costs 15"

# 1. Match every word that starts with an uppercase letter
# 2. Match a run of characters that are neither digits nor whitespace: a negated set
# 3. Match only the characters $ or _ using a set

print(re.findall(r"[A-Z]\w*", receipt))
re.findall(r"[\d\s]+", receipt)
re.findall(r"[$_]", receipt)
```
